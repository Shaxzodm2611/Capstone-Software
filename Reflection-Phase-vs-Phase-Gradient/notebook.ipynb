```python
from matplotlib import pyplot as plt
import math as mt
import numpy as np
# ----- Constants ----- #
p = 20*10**-3 # Seperation between Unit-cells in meta-surface
thetha_i_90 = 90 # standard incident angle
frequency = 2.4*10**9
lambda_0 = 3*10**8 / frequency # Wavelength in free space
```

# Equations

Reflection Angle as a Function of Phase Gradient for $90^\circ$ Incident

$$
\theta_r = \arcsin\!\left(-\frac{\lambda_0}{360^\circ p}\,\Delta\phi\right)
$$

Reflection Angle as a Function of Phase Gradient and Incident Angle

$$
\theta_r = \arcsin\!\left(\sin\theta_i - \frac{\lambda_0}{360^\circ p}\,\Delta\phi\right)
$$

**Plotting Phase Gradient as a Function of Reflection Angle for $90^\circ$ Incident Wave**

$$
\begin{aligned}
360^\circ p\,\sin\theta_r &= -\lambda_0\,\Delta\phi \\
\Delta\phi &= -\frac{360^\circ p}{\lambda_0}\,\sin\theta_r
\end{aligned}
$$


```python
theta_r_deg = np.arange(-90, 91)
theta_r_rad = np.deg2rad(theta_r_deg)

delta_phi_deg = -(360 * p / lambda_0) * np.sin(theta_r_rad)
gradient_rad_m = np.deg2rad(delta_phi_deg) / p

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(theta_r_deg, gradient_rad_m)
ax.set_xlabel("Reflection angle (°)")
ax.set_ylabel("Phase gradient (rad/m)")
ax.set_title("Phase gradient vs. reflection angle at normal incidence")
ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
```

```python
theta_r_deg = np.arange(-90, 91)
theta_r_rad = np.deg2rad(theta_r_deg)
incident_from_surface_deg = np.arange(0, 91, 15)

fig, ax = plt.subplots(figsize=(9, 5))

for angle_deg in incident_from_surface_deg:
    # The steering equation uses an incident angle measured from the normal.
    theta_i_rad = np.deg2rad(90 - angle_deg)
    gradient_rad_m = (2 * np.pi / lambda_0) * (np.sin(theta_i_rad) - np.sin(theta_r_rad))
    ax.plot(theta_r_deg, gradient_rad_m, label=f"{angle_deg}°")

ax.set_xlabel("Reflection angle from normal (°)")
ax.set_ylabel("Phase gradient (rad/m)")
ax.set_title("Phase gradient vs. reflection angle")
ax.legend(title="Incidence angle from surface", ncol=2)
ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
```
